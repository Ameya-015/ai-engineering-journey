```python
import pandas as pd

log = pd.read_csv("data/reading_log.csv", parse_dates=["date"])
log
```

```python
log.isna().sum()
```

```python
log[log.isna().any(axis=1)]
```

```python
dropped = log.dropna(subset=["minutes", "pages_read"])
dropped.shape
```

```python
mins_dropped = log.dropna(subset=["minutes"])
mins_dropped.shape
```

Only one column is dropped which is minutes hence the shape is (6,4)

```python
filled = log.copy().fillna({"minutes": log["minutes"].mean(), "pages_read": log["pages_read"].mean()})
filled
```

```python
dropped["pages_per_min"] = dropped["pages_read"]/dropped["minutes"]
dropped
```

```python
dropped["pages_per_min"].mean()
```

```python
filled["pages_per_min"] = filled["pages_read"]/filled["minutes"]
filled
```

```python
filled["pages_per_min"].mean()
```

- Mean pages/min after dropping: 0.7454761904761904
- Mean pages/min after filling: 0.8552749044823409
- I would choose fill because dropping would erase the row itself which leads to data loss. At least by fill we can retrace why the output is skiewed in certain direction which is still better than not loosing on data
