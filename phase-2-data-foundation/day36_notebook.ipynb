# Day 36 - Movie exploration

```python
import pandas as pd

movies = pd.DataFrame({"title": ["Arrival", "Up", "Dune", "Coco"],
                       "minutes": [116, 96, 155, 105],
                       "year": [2016, 2009, 2021, 2017]})
```

```python
movies
```

```python
movies["decade"] = (movies["year"] // 10) * 10
movies.sort_values("minutes", ascending=False)
```

```python
movies["decade"].value_counts()
```
